```python
import cv2
import numpy as np
import time
import tracemalloc

'''=====圖片路徑更改====='''
image_path = "../Data/quiz1/Zhuang.png"

N = 1000 #執行次數
img = cv2.imread(image_path)
```

```python
# NumPy
tracemalloc.start()
start = time.perf_counter()

for i in range(N):

    B = img[:, :, 0]
    G = img[:, :, 1]
    R = img[:, :, 2]

    gray_numpy = (
        0.114 * B +
        0.587 * G +
        0.299 * R
    ).astype(np.uint8)

end = time.perf_counter()

numpy_total = end - start
numpy_average = numpy_total / N
numpy_current, numpy_peak = tracemalloc.get_traced_memory()

tracemalloc.stop()

# OpenCV
tracemalloc.start()
start = time.perf_counter()

for i in range(N):
    gray_opencv = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

end = time.perf_counter()

opencv_total = end - start
opencv_average = opencv_total / N
opencv_current, opencv_peak = tracemalloc.get_traced_memory()

tracemalloc.stop()

# 結果差異
diff = cv2.absdiff(gray_numpy, gray_opencv)
max_diff = np.max(diff)
mean_diff = np.mean(diff)
different_pixels = np.count_nonzero(diff)
```

```python
# 輸出結果
print("\n========== 執行速度 ==========")

print("NumPy")
print(f"總時間：{numpy_total:.6f} 秒")
print(f"平均時間：{numpy_average:.9f} 秒")

print("\nOpenCV")
print(f"總時間：{opencv_total:.6f} 秒")
print(f"平均時間：{opencv_average:.9f} 秒")

print("\n========== 影像差異 ==========")

print(f"最大像素差異：{max_diff}")
print(f"平均像素差異：{mean_diff:.6f}")
print(f"不同像素數量：{different_pixels}")

print("\n========== 記憶體使用量 ==========")

print("\nNumPy")
print(f"目前記憶體：{numpy_current / 1024**2:.4f} MB")
print(f"峰值記憶體：{numpy_peak / 1024**2:.4f} MB")

print("\nOpenCV")
print(f"目前記憶體：{opencv_current / 1024**2:.4f} MB")
print(f"峰值記憶體：{opencv_peak / 1024**2:.4f} MB")

print("\n========== 灰階影像記憶體 ==========")
print(
    f"NumPy 灰階影像："
    f"{gray_numpy.nbytes / 1024**2:.4f} MB"
)
print(
    f"OpenCV 灰階影像："
    f"{gray_opencv.nbytes / 1024**2:.4f} MB"
)
```

```python
'''=====儲存灰階影像路徑====='''
cv2.imwrite("gray_numpy.jpg", gray_numpy)
cv2.imwrite("gray_opencv.jpg", gray_opencv)
```
